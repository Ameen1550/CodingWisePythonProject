Next, applying IQR Rule


```python
cols = ['age', 'delivery_time_min', 'avg_order_value']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"--- {col} ---")
    print(f"Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
    print(f"Total Outliers: {len(outliers)}\n")
```

Applying Z-Score to these columns

```python
cols = ['age', 'delivery_time_min', 'avg_order_value']

for col in cols:
    mean = df[col].mean()
    std = df[col].std()

    # Calculate Z-score
    z_scores = (df[col] - mean) / std

    # Flag outliers where |Z| > 3
    outliers = df[z_scores.abs() > 3]

    print(f"--- {col} ---")
    print(f"Mean: {mean:.2f}, Std Dev: {std:.2f}")
    print(f"Total Outliers (|Z| > 3): {len(outliers)}\n")
```

Comparision of IQR and Z-Score Values:

1. IQR caught many more outliers (9, 46, and 96) compared to Z-score (5, 4, and 4).

2. IQR uses the middle 50% of the data (median and quartiles), so extreme values will not escape its boundaries—making it ideal for skewed data like order values and delivery times.

3. Use IQR is the Best Method to clean this data with columns as delivery_time_min and avg_order_value, while both methods work well for age since it follows a normal bell-shaped curve.




Now, The best method for data cleaning the data here is to use "Capping Method".

Because we don't want to lose rows from our 2000 customers



```python
cols = ['age', 'delivery_time_min', 'avg_order_value']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    #Here, I am  Capping  values outside the IQR boundaries
    df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

# Now, I am verifying that no values fall outside the IQR boundaries
for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    remaining = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: min={df[col].min():.2f}, max={df[col].max():.2f}, remaining outliers={len(remaining)}")
```

My insughts here:
1. Extreme values innage, delivery_time and avg_order are converted into normal boundaries with 0 outliers.

2. Using ".clip()" kept all 2,000 customer rows intact while removing distortions, so that safely fill in the remaining missing values.

**Now, Imputing "avg_order_value" Column**

```python
df['avg_order_value'] = df['avg_order_value'].fillna(
    df.groupby('city_tier')['avg_order_value'].transform('median')
)

print("Missing values left in avg_order_value:", df['avg_order_value'].isna().sum())
```

This fills all 108 remaining missing values in avg_order_value and leaves 0 missing values across the entire DataFrame.

1. Global Fill (Overall Median) means, it replaces all missing values with a single dataset-wide number (315.5). While simple, it ignores regional differences and artificially flattens spending patterns across different city tiers.

2. Group-wise Fill (Grouped by city_tier): Fills missing values using the specific median of each customer's city_tier (Tier 1, Tier 2, or Tier 3).

3. Because order values vary depending on city size and purchasing orders, "group-wise fill" is much more accurate and has realistic spending within each tier.

Explain why the order of mean, median, mode flips between
left- and right-skewed data

It flips beacause The Mean gets pulled toward extreme values (the long tail), while the Mode stays stuck at  highest peak,and the Median stays in the middle.

*  In Right Skwewd Data:
Order (Left to Right): Mode < Median < Mean

therefore,unusally  high numbers on the right act like heavy weights and pull the Mean to  right, while the Mode stays at main peak on the left


*  In Left skewed Data:
Order (Left to Right): Mean < Median < Mode

That's why,whenever unusually low numbers on left  are there, they drag the Mean down to the left, while the Mode stays at the main peak on the right


*  Meanwhile in Symmetric Data:
Order: Mean = Median = Mode

Because both sides are balanced, nothing pulls the Mean/ average  away from the peak, so all three stays at middle / centre of graph. Hence Balanced

## **Task 5**

Label/ Ordinal Encoding:

Here in given Membershiip, there is a ranking like :
Basic -> Silver -> Gold -> Platinum
So we replace them with 0, 1, 2, 3

```python
print(df['membership'].unique())
membership_map = {'Basic': 0, 'Silver': 1, 'Gold': 2, 'Platinum': 3}
df['membership_encoded'] = df['membership'].map(membership_map)
```

One Hot Encoding:

So, Given as device_type and acquisition channel are categories , so no ranking here.

So, we create seperate "0's and 1's  columns" for each category

And here to avaoid confusion of duplicates, we drop the first column with "drop_first = true"



```python
dummies = pd.get_dummies(df[['device_type', 'acquisition_channel']], drop_first=True)
df = pd.concat([df, dummies], axis=1)
```

Target Encoding :

So, here instead of turning city into 10 separate columns, we replace each city's name with the average annual_spend of customers living in that city.

```python
df['city_target_encoded'] = df.groupby('city')['annual_spend'].transform('mean')
```

Why label-encoding city alphabetically would be misleading:

Assigning numbers alphabetically
E.g., Bengaluru = 0,
Indore = 1,
 Mumbai = 2

 This mkaes the model into thinking as Mumbai > Indore > Bengaluru,
 even though city names have no mathematical order


Why one-hot encoding on 10 cities is "wide"?

One-hot encoding creates a brand-new column for almost every city.
Therefore , this  makes the dataset unnecessarily wide and maximmum filled with zeros

Risk of Target Encoding (Data Leakage):

Beacause it is like cheating on a test because we use the final answer (annual_spend) to create the question (city_target_encoded).
The model memorizes the training answers instead of truly learning, so it performs poorly on new customers.

## **Task 7: Pareto Distribution and the 80/20 Rule**

Sort customers by annual_spend , compute cumulative revenue share

```python
pareto_df = df.sort_values(by='annual_spend', ascending=False).copy()

total_revenue = pareto_df['annual_spend'].sum()
pareto_df['cumulative_revenue'] = pareto_df['annual_spend'].cumsum()
pareto_df['cum_revenue_pct'] = (pareto_df['cumulative_revenue'] / total_revenue) * 100

pareto_df['customer_rank'] = range(1, len(pareto_df) + 1)
pareto_df['cum_customer_pct'] = (pareto_df['customer_rank'] / len(pareto_df)) * 100

pareto_df[['customer_id', 'annual_spend', 'cum_customer_pct', 'cum_revenue_pct']].head(10)

```

Here,
Just 0.5% of customers (the top 10 people) bring in over 12% of total revenue.

A tiny group of high spenders drives a massive share of earnings, confirming a strong 80/20 Pareto pattern.

Next,
Plot a Pareto chart (bars for spend, line for cumulative %) or a cumulative-share curve.


```python
plt.plot(pareto_df['cum_customer_pct'], pareto_df['cum_revenue_pct'], color='red')
plt.axhline(80, color='gray', linestyle='--')
plt.xlabel('Cumulative Customers (%)')
plt.ylabel('Cumulative Revenue (%)')
plt.title('Cumulative Revenue Share')
plt.show()
```

Here, we get:
The red curve shoots up at the start, showing that small amount of top customers brings most revenue.

The curve crosses the 80% dashed line, proves a strong 80/20 Pareto pattern.

After that point, the curve is almost flat only, meaning the remaining majority of customers add very little amount of earnings.

Find what percentage of customers contribute 50% and 80% of revenue

```python
count_50 = len(pareto_df[pareto_df['cum_revenue_pct'] < 50]) + 1
count_80 = len(pareto_df[pareto_df['cum_revenue_pct'] < 80]) + 1

total_customers = len(pareto_df)
pct_50 = (count_50 / total_customers) * 100
pct_80 = (count_80 / total_customers) * 100

print(pct_50, "% of customers contribute 50% of revenue")
print(pct_80, "% of customers contribute 80% of revenue")
```

Compare the top 20% share with the classic 80/20 idea.

```python
top_20 = pareto_df[pareto_df['cum_customer_pct'] <= 20]

top_20_revenue = top_20['annual_spend'].sum()
total_revenue = pareto_df['annual_spend'].sum()
top_20_share = (top_20_revenue / total_revenue) * 100

print("Top 20% of customers brings ", round(top_20_share, 2), "% of revenue")
```

here, the insights got are:
The top 20% of customers bring in 60.61% of total revenue, while 43.1% of customers are needed to reach 80%.

Spending is heavily skewed toward top buyers, but it is less extreme than the classic 80/20 rule—meaning mid-tier customers still have big share of earnings.

## **Task 8: PDF and CDF**

Plot the PDF (KDE or normalised histogram) of delivery_time_min

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['delivery_time_min'], stat='density', kde=True, color='green')
plt.title('PDF of Delivery Time')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Density')
plt.show()
```

here, stat='density' normalises the bars so the total area under histogram is equals to  1


Plot the empirical CDF ( np.sort + np.arange(1, n+1)/n )

```python
x = np.sort(df['delivery_time_min'])
n = len(x)
y = np.arange(1, n + 1) / n

plt.figure(figsize=(8, 4))
plt.plot(x, y, color='blue')
plt.title('Empirical CDF of Delivery Time')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Cumulative Probability')
plt.show()
```

np.sort() lines up all 2,000 delivery times from fastest to slowest on the x-axis.

np.arange(1, n + 1) / n creates evenly spaced steps from 0 to 1 on the y-axis

From the CDF, answer:

What is the probability a delivery takes ≤ 30 minutes?

What is the probability a delivery takes > 45 minutes?

What time corresponds to the median and the 95th percentile?

```python
prob_30 = (df['delivery_time_min'] <= 30).mean() * 100
print("Probability <= 30 min:", round(prob_30, 2), "%")


prob_45 = (df['delivery_time_min'] > 45).mean() * 100
print("Probability > 45 min:", round(prob_45, 2), "%")

median_time = df['delivery_time_min'].quantile(0.50)
print("Median (P50):", median_time, "minutes")


p95_time = df['delivery_time_min'].quantile(0.95)
print("95th Percentile (P95):", p95_time, "minutes")
```

Explain in plain language: PDF = how dense values are around a point; CDF = share of

CDF: Shows what percentage of employees earn ₹40,000 or less.

My simple trick:
PDF = Where is the crowd?
